```python
# Crear un sistema de usuarios con contraseñas y emails

usuarios = {
    1: {
        "nombre": "Juan",
        "contraseña": "abc123!",
        "email": "test@test.com"
    },
    2: {
            "nombre": "Marta",
            "contraseña": "1234",
            "email": "marta@test.com"
        },
    3: {
            "nombre": "Joaquin",
            "contraseña": "666",
            "email": "joa_profe@test.com"
        },
    4: {
            "nombre": "Ana",
            "contraseña": "kjabsdk",
            "email": None
        }
}

# Logica de programacion
mi_usuario = input("Pon tu usuario: ")
mi_contraseña = input("Pon tu contraseña: ")
usuario_encontrado = False
for usuario in usuarios.values():
    if mi_usuario.lower() == usuario["nombre"].lower():
        usuario_encontrado = True
        if mi_contraseña == usuario["contraseña"]:
            print("Usuario y contraseña correctos!")
            break
        else:
            print("La contraseña no fue correcta!")
        if not usuario["email"]:  # Si no tenemos un email
            nuevo_email = input("Tu usuario no tiene email, crea uno!\n")
            if nuevo_email.count("@") == 1:  # todo bien!
                usuario["email"] = nuevo_email
            else:
                print("Ese email no fue valido!")

if not usuario_encontrado:
    print("El usuario no se encontro")
```

```python
# Vamos a hacer un menu

menu = """
### Bienvenido al codigo de Tajamar! ###

1. Leer datos
2. Nueva entrada
3. Actualizar entrada
4. Salir
"""

print(menu)
comportamiento = {
    1: "leer datos",
    2: "crear entrada",
    3: "actualizar una entrada",
    4: "salir"
}

eleccion = None
while not eleccion:
    eleccion_usuario = int(input("\n¿Que opcion tomas?\n"))
    if eleccion_usuario in comportamiento:
        eleccion = eleccion_usuario
        break
    else:
        print("Esa eleccion no es valida!")

print(f"Has elegido la opcion {eleccion} ---> {comportamiento[eleccion]}")
 
```

```python
#Ejercicio
# Guardar los primeros 100 números primos en una lista

# primos del 0 al 100
prime_list = []

for i in range(0,100):
    prime = True
    for j in range(2,i):
        if i % j == 0:
            prime = False
            break
    if not prime:
        continue
    else:
        prime_list.append(i)

print(prime_list)   
```

```python
# Monitor de servidores

servidores = {
    "web-1": {"cpu": 40, "ram": 800, "health": True},  # health es encendido o apagado
    "web-2": {"cpu": 85, "ram": 2000, "health": False},
    "db-1": {"cpu": 99, "ram": 80, "health": True},
    "db-2": {"cpu": 20, "ram": 10000, "health": False},
    "web-3": {"cpu": 5, "ram": 512, "health": False},
    "front-1": {"cpu": 97, "ram": 2048, "health": True},
}

# Detectar:
#   CPU superior a 80%  |  RAM menos de 512  |  Criticos estan encendidos y cumplen ambas

servidores_criticos = []
for servidor, valores_servidor in servidores.items():
    cpu_alta = 80
    memoria_baja = 512
    estado_servidor = 0  # 1 esta mal, 2 esta critico

    if valores_servidor["cpu"] >= cpu_alta:
        estado_servidor += 1

    if valores_servidor["ram"] <= memoria_baja:
        estado_servidor += 1

    if valores_servidor["health"]:
        if estado_servidor <= 0:
            print(f"El servidor {servidor} esta estable")
        elif estado_servidor == 1:
            print(f"El servidor {servidor} esta en mal estado")
        elif estado_servidor == 2:
            servidores_criticos.append(servidor)
            print(f"El servidor {servidor} esta en estado critico!")
    else:
        print(f"El servidor {servidor} esta apagado")

print("-" * 20)  # separador en el menu

print(f"Tenemos {len(servidores_criticos)} servidores criticos, que son {servidores_criticos}")
 
```

```python
# Sistema de autenticacion de usuarios

usuarios = {
    "joaquin": {
        "rol": "admin",
        "permisos": ["lectura", "escritura", "devops", "admin"],
        "activo": True
    },
    "maria": {
            "rol": "devops",
            "permisos": ["lectura", "devops"],
            "activo": True
    },
    "juan": {
            "rol": "developer",
            "permisos": ["lectura", "escritura"],
            "activo": True
    },
    "silvia": {
            "rol": "sales",
            "permisos": ["lectura"],
            "activo": False
    },
}

usuarios["silvia"]["activo"] = True
usuarios.update({"pedro": {"rol": "developer", "permisos": ["lectura", "escritura"], "activo": True}})
usuarios["juan"]["activo"] = False

permiso_a_consultar = input("Que permiso quieres consultar?\n")

usuarios_con_permisos = []
for usuario, info in usuarios.items():
    permisos = info["permisos"]
    activo = info["activo"]
    if permiso_a_consultar in permisos and activo:
        usuarios_con_permisos.append(usuario)

print(f"Los usuarios con permisos de {permiso_a_consultar} son {len(usuarios_con_permisos)}\n{usuarios_con_permisos}")
```
